```python
import pandas as pd 

df1 = pd.read_csv('datasets/ChurnPredictionData/churn-bigml-20.csv')
df1.head()
```

```python
df2 = pd.read_csv('datasets/ChurnPredictionData/churn-bigml-80.csv')
df2.head()
```

```python
print('Dataset 1 Columns:', df1.columns)
print('Dataset 2 Columns:', df2.columns)

df1.shape, df2.shape
```

```python
df1.info()
```

```python
df2.info()
```

```python
# Combining both datasets
df = pd.concat([df1, df2], ignore_index=True)
df.head(25)
```

```python
# checking shape of combined dataset
df.shape
```

```python
# checking for missing values
print(df.isnull().sum())
```

```python
# checking for duplicates
df.duplicated().sum()
```

```python
# Checking values for 'Churn' column
df['Churn'].value_counts()
```

```python
# Convert 'Churn' column to binary
df['Churn'] = df['Churn'].astype(int)
print(df['Churn'].value_counts())
```

```python
# Convert 'International plan' and 'Voice mail plan' to binary
df['International plan'] = df['International plan'].map({'Yes': 1, 'No': 0})
print(df['International plan'].value_counts())
    
df['Voice mail plan'] = df['Voice mail plan'].map({'Yes': 1, 'No': 0})
print(df['Voice mail plan'].value_counts())
```

```python
# Calculating overall churn rate
churn_rate = df['Churn'].value_counts(normalize=True)
print("Overall Churn Rate:\n", churn_rate)
```

```python
# Analyzing churners and Non-churners

# See if they make more service calls
print(df.groupby('Churn')['Customer service calls'].mean())

# See if they have higher daily charges
print(df.groupby('Churn')['Total day charge'].mean())
```

```python
# Looking at statistics for churn column
churn_rate.describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Visualizing churn distribution
sns.countplot(x='Churn', data=df)
plt.title('Churn Distribution')
plt.show()
```

```python
# Visualizing impact on Customer service calls
sns.countplot(x='Churn', hue='Customer service calls', data=df)
plt.title('Customer Service Calls by Churn Status')
plt.show()
```

```python
# Visualizing impact of international plan
sns.countplot(x='Churn', hue='International plan', data=df)
plt.title('International Plan by Churn Status')
plt.show()
```
