```python
# Giving columns names
column_names = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J', 'K', 'L', 'M', 'N']
df = pd.read_csv('datasets/housePrediction.csv', sep='\s+', names=column_names)
print(df.head())
```

```python
import matplotlib.pyplot as plt

# Plotting Distribution of target variable (N - house prices)
plt.figure(figsize=(8, 5))
plt.hist(df['N'], bins=30, edgecolor='black')
plt.title('Distribution of House Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

```python
# Checking for correlation between features and house prices
plt.figure(figsize=(10, 6))
# Select a few important features for correlation plot
important_features = ['A', 'B', 'C', 'F', 'G', 'N']  # N is the target (house price)
correlation_data = df[important_features].corr()

import seaborn as sns
sns.heatmap(correlation_data, annot=True, cmap='coolwarm', center=0)
plt.title('Feature Correlation with House Prices')
plt.show()
```

```python
# Scatter plot - Room count vs House Price
plt.figure(figsize=(8, 5))
plt.scatter(df['G'], df['N'], alpha=0.6)
plt.title('Number of Rooms vs House Price')
plt.xlabel('Number of Rooms (G)')
plt.ylabel('House Price (N)')
plt.show()
```
